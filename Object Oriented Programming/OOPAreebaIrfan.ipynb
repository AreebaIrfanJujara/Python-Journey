```python
class Animal:
    def eat(self):
        print("Animal eating")

class Mammal(Animal):
    def walk(self):
        print("Mammal walking")

class Dog(Mammal):
    def bark(self):
        print("Dog barking")

# Calling methods
d = Dog()
d.eat()
d.walk()
d.bark()
```

```python
class Person:
    def details(self):
        print("Person details")

class Employee(Person):
    def work(self):
        print("Employee working")

class Manager(Employee):
    def manage(self):
        print("Manager managing")

# Calling
m = Manager()
m.details()
m.work()
m.manage()
```

```python
class Vehicle:
    def move(self):
        print("Vehicle moving")

class Car(Vehicle):
    def fuel(self):
        print("Car using fuel")

class SportsCar(Car):
    def speed(self):
        print("SportsCar very fast")

# Calling
s = SportsCar()
s.move()
s.fuel()
s.speed()
```

```python
class LivingThing:
    def breathe(self):
        print("Breathing")

class Animal(LivingThing):
    def eat(self):
        print("Eating")

class Cat(Animal):
    def meow(self):
        print("Meowing")

# Calling
c = Cat()
c.breathe()
c.eat()
c.meow()
```
